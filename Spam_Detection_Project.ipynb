# Task 4 - Email/SMS Spam Detection with Machine Learning
### Objective: Build an NLP binary classifier to distinguish spam from ham.
**Tech Stack:** Python, pandas, scikit-learn, TF-IDF, Naive Bayes, Logistic Regression

```python
import pandas as pd
import re
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Load dataset from storage path discovered on device
file_path = '/storage/emulated/0/Documents/spam zip/sms+spam+collection (1) (1)/SMSSpamCollection'
df = pd.read_csv(file_path, sep='\t', names=['label', 'message'])
print(df['label'].value_counts())
```

## Feature Extraction Explanation
**TF-IDF (Term Frequency-Inverse Document Frequency)** measures how important a specific word is to an individual text message relative to the frequency of that word across the entire dataset collection.

```python
# Text preprocessing pipeline
df['clean_message'] = df['message'].astype(str).apply(lambda x: re.sub(r'[^\w\s]', '', x.lower()))

# Splitting data into training and testing partitions
X_train, X_test, y_train, y_test = train_test_split(df['clean_message'], df['label'], test_size=0.2, random_state=42)

# Vectorize utilizing TF-IDF Vectorizer with stopword removal
tfidf = TfidfVectorizer(stop_words='english')
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)
```

```python
# Training and Evaluating Multinomial Naive Bayes
nb_model = MultinomialNB().fit(X_train_vec, y_train)
print('--- Multinomial Naive Bayes Report ---')
print(classification_report(y_test, nb_model.predict(X_test_vec)))
```

```python
# Training and Evaluating Alternative Classifier (Logistic Regression)
lr_model = LogisticRegression().fit(X_train_vec, y_train)
print('--- Logistic Regression Report ---')
print(classification_report(y_test, lr_model.predict(X_test_vec)))
```

## Discussion
**Why is recall important for spam detection?**

High recall ensures that our model filters out and catches the vast majority of malicious spam items. In real-world consumer settings, however, precision for benign email ('ham') is maximized to ensure critical messages are not accidentally redirected to spam archives.
